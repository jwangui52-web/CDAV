```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/student_performance.csv")
```

```python
df.head(11)
```

```python
grade=df["FinalGrade"]
print(grade)
```

```python
grade.mean()
```

```python
grade.mode()
```

#no mode

```python
#Quantiles.....>Q1,Q2,Q3
print(f"Q1: {grade.quantile(0.25)}")
print(f"Q2: {grade.quantile(0.50)}")
print(f"Q3: {grade.quantile(0.75)}")



```

```python
#Deciles
print(f"D1: {grade.quantile(0.10)}")
print(f"D2: {grade.quantile(0.50)}")
print(f"D3: {grade.quantile(0.30)}")



```

```python
#percentiles
print(f"Q1: {grade.quantile(0.01)}")
print(f"Q2: {grade.quantile(0.50)}")
print(f"Q3: {grade.quantile(0.30)}")



```

```python
#range
maximum=grade.max()
minimum=grade.min()
grade_range=maximum-minimum
grade_range
```

```python
Q3=grade.quantile(0.75)
Q1=grade.quantile(0.25)
IQR=Q3-Q1
IQR
```

```python
#variance and std
var=grade.var()
std=grade.std()
var
```
