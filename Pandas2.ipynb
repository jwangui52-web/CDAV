```python
df.columns=(
    df.columns.str.strip()    #remove leading /trailing spaces
    .str.lower()              #make everything lowercase
    .str.replace(' ','_')     #spaces inside names become underscores
)
```

```python
#clean the county column
print('unique counties BEFORE cleaning')
```

```python
df.pivot_table(
    index="county",
    columns="status",
    values="amount_paid",
    aggfunc="sum",
    )
```

```python
import matplotlib.pyplot as plt
sales=df.groupby("county")["amount_paid"].sum().sort_values()
sales.plot(kind="barh")
plt.show()
```

```python
df["amount_paid"].plot(kind="box")
df.boxplot(column="amount_paid",by="county")
```

```python
ax = sales.plot(kind="bar", color="#B3A6B")
ax.set_title("Total sales by county,Jan-Jun 2026")
ax.set_ylabel("sales(KES)")
ax.set_xlabel("")
plt.xticks(rotation=40,ha="right")
plt.tight_layout()
```

```python
import seaborn as sns
#count of transactions per county,one line
sns.countplot(data=df,x="county")
```

```python
#average amount per county,with error bars
sns.barplot(data=df,x="county",y="amount_paid")
```

```python
#saving a chart to file
ax=sales.plot(kind="bar")
ax.set_title("Total sales by county")
plt.savefig("sales_by_county.png",dpi=150,bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt
df["amount_paid"].plot(kind="hist", bins=30, title="sales sizes")
df.groupby("county")["amount_paid"].sum().plot(kind="barh")
df.plot(kind="scatter",x="rainfall_mm",y="amount_paid")

```

```python
import matplotlib.pyplot as plt

# Convert 'txn_date' to datetime objects, coercing errors to NaT
df['txn_date'] = pd.to_datetime(df['txn_date'], format='%d/%m/%Y', errors='coerce')

# Extract month from 'txn_date'
df['month'] = df['txn_date'].dt.month

# Define MONTHS for reindexing
MONTHS = range(1, 13)  # Assuming months are represented as integers 1-12

df.groupby("month")["amount_paid"].sum().reindex(MONTHS).plot(kind="line")
plt.show()
```
