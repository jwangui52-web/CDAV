```python
def stat(num1,num2):
  return (num1+num2), (num1+num2)/2
```

```python
ustat=np.frompyfunc(stat,2,2)
arr_1=np.array([1,2,3])
arr_2=np.array([3,4,5])
ustat(arr_1,arr_2)
```

```python
sum_of_arrs,mean_of_arrs=ustat(arr_1,arr_2)
```
