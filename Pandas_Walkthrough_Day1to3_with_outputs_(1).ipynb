## 2. Filtering  *(Day 2)*

Nobody analyses 1000 rows. They analyse the rows that matter. Filtering is how you
keep only those. It happens in two steps: **ask a question**, then **use the answer**.

**Step 1 — ask a question.** A comparison makes a column of True/False, one per row.
This is called a *boolean mask*.

```python
df['Rainfall mm'] > 1000
```

**Step 2 — use the mask** to keep only the True rows.

```python
df[df['Rainfall mm'] > 1000].head()
```

**Two conditions at once.** Use `&` for *and*, `|` for *or*.
Every condition needs its **own brackets** — this is the number-one beginner error.

```python
df[(df['Rainfall mm'] > 1000) & (df['Status'] == 'P')].head()
```

**Before filtering a text column, see what is actually in it.** `value_counts()` lists every
value and how often it appears.

```python
df['County '].value_counts()
```

Look at that result. **Nairobi is spelled several different ways** — `Nairobi`, `nairobi`,
`NAIROBI`, and one with a space. To pandas these are *different counties*. If we counted
sales per county right now, Nairobi would be split four ways and every answer would be wrong.

We will fix this in the cleaning section. For now, just notice the problem is real and silent.

## 3. Cleaning  *(Day 2 — the heart of the week)*

This is what “data cleaning” actually means: taking a broken table and making every column
say what it is supposed to say. We do it in a set order.

### 3a. Fix the column names first

Names with spaces and capitals are a trap — ` Amount Paid ` will cause a `KeyError` you
cannot see. Clean them all in one line, so everything after this is easy to type.

```python
df.columns          # the messy names, before
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df = df.rename(columns={'amount_paid': 'amount'})
df.columns          # clean names, after
```

From here on we can type `df['amount']` and `df['county']` — no spaces, no surprises.

### 3b. Find the missing values

`isna().sum()` counts the gaps in every column. Add this to your first-look habit.

```python
df.isna().sum()
```

### 3c. Find duplicate rows

How many rows are exact repeats?

```python
df.duplicated().sum()
```

And how many **distinct transaction IDs** are there, versus total rows? If these differ,
some IDs appear more than once.

```python
print('rows:', len(df))
print('distinct txn_id:', df['txn_id'].nunique())
```

We will remove duplicates later, once the rest is clean — so we only decide once.

### 3d. Rescue the amount column

`amount` came in as **text** because some values have commas or “KES”. Look:

```python
df['amount'].head(10)
```

The fix is always **strip the junk first, then convert**. `errors='coerce'` turns anything
it cannot read (like `-` or `NA`) into `NaN`, which we already know how to handle.

```python
clean = (df['amount'].astype(str)
                     .str.replace(',', '', regex=False)
                     .str.replace(' KES', '', regex=False)
                     .str.strip())
df['amount'] = pd.to_numeric(clean, errors='coerce')
df['amount'].dtype        # now a real number
```

How many values could not be converted (they became `NaN`)?

```python
df['amount'].isna().sum()
```

Some amounts are **negative**, which is impossible for a sale. Set those to missing — we do
not know the true value, so we should not guess it. This is the correct way to change part of
a table: `df.loc[condition, column] = value`.

```python
print('negative amounts:', (df['amount'] < 0).sum())
df.loc[df['amount'] < 0, 'amount'] = np.nan
```

Now that `amount` is numeric, it finally appears in `describe()` — it was invisible before.

```python
df['amount'].describe()
```

---
## 4. Cleaning text  *(Day 3)*

Now we kill the four Nairobis.

The `.str` accessor applies a text method to **every row at once**. Chain the fixes:
remove the sneaky hidden character, strip the spaces, then Title-Case everything.

```python
df['county'] = (df['county'].str.replace('\u200b', '', regex=False)  # a hidden 'zero-width' space
                            .str.strip()
                            .str.title())
df['county'].value_counts()
```

**One line, and the four Nairobis became one.** Every count and chart from here on will be correct.

### Split a column into two

`full_name` is `“Surname, First”`. Split it on the comma into two real columns.

```python
df['full_name'].head()
```

```python
parts = df['full_name'].str.split(',', expand=True)
df['surname']    = parts[0].str.strip()
df['first_name'] = parts[1].str.strip()
df[['surname', 'first_name']].head()
```

## 5. Dates  *(Day 3)*

Right now the dates are just **text** — you cannot sort them properly or do arithmetic.

```python
df['txn_date'].dtype        # object = text
```

Convert to real dates. These are **DD/MM/YYYY** (Kenyan style), so we must say `dayfirst=True`
or pandas reads the day as the month. Bad dates become `NaT` (the date version of `NaN`).

```python
df['txn_date']  = pd.to_datetime(df['txn_date'],  dayfirst=True, errors='coerce')
df['paid_date'] = pd.to_datetime(df['paid_date'], dayfirst=True, errors='coerce')
df['txn_date'].dtype        # now a real date
```

Now we can pull the pieces out with `.dt`, just like `.str` for text.

```python
df['month']   = df['txn_date'].dt.month_name()
df['weekday'] = df['txn_date'].dt.day_name()
df[['txn_date', 'month', 'weekday']].head()
```

And do **date arithmetic** — how many days did customers take to pay?

```python
df['days_to_pay'] = (df['paid_date'] - df['txn_date']).dt.days
df['days_to_pay'].describe()
```

---
## 6. Building new columns  *(Day 3)*

Most new columns are just maths or a comparison — no fancy tools needed.

```python
df['is_large'] = df['amount'] > 1000                     # a True/False flag
```

```python
df['size'] = np.where(df['amount'] > 1000, 'big', 'small')  # two outcomes
```

Tidy the status codes into readable words. `replace` swaps values; mapping `'?'` to `None`
turns the unknowns into proper missing values.

```python
df['status'] = df['status'].replace({'P': 'paid', 'U': 'unpaid', '?': np.nan})
df['status'].value_counts(dropna=False)
```

---
## 7. Duplicates and missing values — the decisions  *(Day 2)*

We saved these for last, so we decide once, on clean data.

**Duplicates:** a repeated `txn_id` is a genuine duplicate here, so keep the first of each.

```python
before = len(df)
df = df.drop_duplicates(subset=['txn_id']).copy()
print('removed', before - len(df), 'duplicate rows')
```

**Missing values — a judgement call, not a rule.** Here we:
- **drop** rows with no `amount` (we cannot analyse a sale with no value), and
- **fill** missing rainfall with the median (a sensible, honest default).

The important part is that we can *explain* each choice — that is the graded skill.

```python
df = df.dropna(subset=['amount'])
df['rainfall_mm'] = df['rainfall_mm'].fillna(df['rainfall_mm'].median())
df.isna().sum()
```

> **Write it down.** In a real project you would add a note like:
> *“Dropped rows with missing amount (cannot value the sale); filled missing rainfall with the
> median; removed duplicate transaction IDs.”* The code is not enough on its own.
