#Matplotlib Deep -dive on line charts


```python
rng=np.random.default_rng(42)
n=12

df=pd.DataFrame({
    "month":pd.date_range("2024-01-01", periods=n, freq=("MS")).strftime("%b"),
    "sales": rng.integers(50,100,n),
    "cost": rng.integers(20,60,n),
    "region": rng.choice(["East","West"],n),
})
df["profit"]=df["sales"]-df["cost"]
df.head()
```

```python
#simple line plot
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"])
plt.show()
```

```python
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"]),
ax.plot(df["month"],df["cost"]),
ax.plot(df["month"],df["profit"]),
plt.show()


```

```python
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"], color="crimson")
ax.plot(df["month"],df["cost"], color="steelblue")
ax.plot(df["month"],df["profit"], color="seagreen")
plt.show()


```

```python
fig,ax=plt.subplots()
ax.plot(df["month"], df["sales"], linestyle="-", linewidth=2, marker="o")

ax.plot(df["month"], df["cost"], linestyle="-",  linewidth=2, marker="s")

ax.plot(df["month"], df["profit"],linestyle="-", linewidth=2, marker="^")

plt.show()


```

```python
fig,ax=plt.subplots()
ax.plot(df["month"], df["sales"], linestyle="-", linewidth=2, marker="o")

ax.plot(df["month"], df["cost"], linestyle="-",  linewidth=2, marker="s")

ax.plot(df["month"], df["profit"],linestyle="-", linewidth=2, marker="^")
ax.set_title("line chart")
ax.set_xlabel("month")
ax.set_ylabel("sales")
ax.legend("sales,cost,profit")
plt.show()


```

```python
#Clustered Bar
x=np.arange(len(df))
fig,ax= plt.subplots()
ax.bar(x-0.2,df["sales"],0.4, label="sales", color= "darkblue")

ax.bar(x+0.2,df["cost"],0.4,  label="cost", color= "rosybrown")

ax.set_xticks(x,df["month"])
ax.set_ylabel("sales")
ax.set_xlabel("month")
ax.set_title("sales and cost",fontsize="15",fontweight="bold")

ax.legend()
plt.show()
```
